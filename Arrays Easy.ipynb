```python
arr = [1, 2, 4, 6, 7, 5]

def getElements(arr):
    max_val = max(arr)
    min_val = min(arr)

    second_small = float('inf')
    second_largest = float('-inf')

    for i in range(len(arr)):
        if arr[i] > second_largest and arr[i] != max_val:
            second_largest = arr[i]

        if arr[i] < second_small and arr[i] != min_val:
            second_small = arr[i]

    print("Second Smallest : ", second_small)
    print("Second Largest : ", second_largest)


getElements(arr)
```

```python
# Check if an Array is Sorted
arr = [1,2,3,4,5]

def checkSorted(arr):
    for i in range(1, len(arr)):
        if arr[i] < arr[i-1]:
            return False
    return True

if (checkSorted(arr)):
    print("True")
else:
    print("False")
    
```

```python
arr = [1, 1, 1, 2, 2, 3, 3, 3, 3, 4, 4]

def removeDuplicates(arr):
    i = 0

    for j in range(1, len(arr)):
        if arr[j] != arr[i]:
            i += 1            # Move the wall one step right
            arr[i] = arr[j]   # Overwrite the duplicate with the new unique value

    return i + 1

unique_count = removeDuplicates(arr)

print("Number of unique elements:", unique_count)
print("The array now looks like:", arr[:unique_count])
```

```python
# Left rotate the Array by One
nums = [1,2,3,4,5]

def rotateByOne(nums):
    temp = nums[0]
    
    for i in range(1, len(nums)):
        nums[i-1] = nums[i]
    nums[-1] = temp

    return nums


rotateByOne(nums)

print(nums)


    


    
```

```python
# Left rotate the Array by K places

nums = [1,2,3,4,5,6,7]
k = 3

def rotateByOne(nums, k):
    n = len(nums)
    
    k %= n
    
    def reverse(left, right):
        while left < right:
            nums[left], nums[right] = nums[right], nums[left]
            left += 1
            right -= 1
    
    # 1. Reverse the entire array
    reverse(0, len(nums) - 1)
    # 2. Reverse the first k elements
    reverse(0, k - 1)
    # 3. Reverse the remaining n - k elements
    reverse(k, len(nums) - 1)

    
```

```python
# Union of Two Sorted Arrays

arr1= [3, 4, 6, 7, 9, 9]
arr2= [1, 5, 7, 8, 8]

def find_union(arr1, arr2):
    i = 0
    j = 0
    union = []
    # Make sure this while loop continue until the end
    while i < len(arr1) and j < len(arr2):

        # if the 1st array, is smaller than 2nd array then append
        if arr1[i] < arr2[j]:
            # make sure it shouldn't be duplicate
            if not union or union[-1] != arr1[i]:
                union.append(arr1[i])
            i+=1

        # same applies for this one also
        elif arr2[j] < arr1[i]:
            if not union or union[-1] != arr2[j]:
                union.append(arr2[j])
            j+=1

        # if both are same then you append the union value.
        else:
            if not union or union[-1] != arr1[i]:
                union.append(arr1[i])
            i += 1
            j += 1

    
    # this part is basically adding remaining part of Array 1
    while i < len(arr1):
        if union[-1] != arr1[i]:
            union.append(arr1[i])
        i += 1

    # same applies for the array 2 also
    while j < len(arr2):
        if union[-1] != arr2[j]:
            union.append(arr2[j])

        j+=1

    return union


print(find_union(arr1, arr2))
```
